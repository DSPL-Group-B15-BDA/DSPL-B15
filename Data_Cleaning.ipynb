#luxury_sales, fresh_sales, dry_sales

Median imputation is relatively simple to implement and computationally efficient. Sales data often falls under a skewed distribution, where the majority of sales might fall within a range with a few exceptionally high sales. The median is a better representation of the central tendency in skewed data compared to the mean.

```python
test_check = pd.read_csv('test.csv')
test_check[test_check['Customer_ID'].isin([10364,4059,28662,28664])]
```

```python
#convert textual figures into numbers in test csv
test_df.loc[test_df['Customer_ID']== 10364, 'luxury_sales'] = 760
test_df.loc[test_df['Customer_ID']== 4059, 'luxury_sales'] = 815
test_df.loc[test_df['Customer_ID']== 28662, 'fresh_sales'] = 9000
test_df.loc[test_df['Customer_ID']== 28664, 'dry_sales'] = 3700
```

*Handling missing values*

```python
#Median Imputation for luxury_sales by taking the median sales for the relevant outlet_city in test csv
median_luxury_sales = test_df.groupby('outlet_city')['luxury_sales'].median()

#Fill missing values
test_df['luxury_sales'] = test_df.apply(
    lambda row: median_luxury_sales[row['outlet_city']] if pd.isnull(row['luxury_sales']) else row['luxury_sales'],
    axis=1)

#check for missing values in the luxury_sales column
missing_luxury_sales = test_df[test_df['luxury_sales'].isnull()]
missing_luxury_sales
```

```python
#missing values as a percentage
missing_values_percentage = (test_df.isnull().sum() / len(test_df)) * 100
missing_values_percentage
```

```python
#Ensure no negative values in train csv
df = train_df[(train_df['luxury_sales'] >= 0) & (train_df['fresh_sales'] >= 0) & (train_df['dry_sales'] >= 0)]
```
